# 5. Evaluation
Create a tiny repeatable evaluation dataset.

```python
tests = [
 {'question':'Which plan has audit logs?', 'expected':'Business'},
 {'question':'What is the new-customer refund window?', 'expected':'fourteen'},
 {'question':'What is the Priority 1 response time?', 'expected':'thirty minutes'},
]
```

```python
def keyword_score(answer: str, expected: str) -> int:
    return int(expected.lower() in answer.lower())
```

```python
sample_answers = [
 'The Business plan includes audit logs.',
 'The refund window is fourteen calendar days.',
 'Priority 1 receives a response within thirty minutes.'
]
scores=[keyword_score(a,t['expected']) for a,t in zip(sample_answers,tests)]
sum(scores)/len(scores)
```

## Important
Keyword scoring is only a baseline. Evaluate retrieval, groundedness, correctness, relevance, and format separately.
