## 1. Import memory components

````python
from langchain.agents import create_agent
from langchain.chat_models import init_chat_model
from langgraph.checkpoint.memory import InMemorySaver
from dotenv import load_dotenv

load_dotenv()
MODEL = "openai:gpt-4o-mini"
````

## 2. Create an agent with a checkpointer

````python
agent = create_agent(
    model=init_chat_model(MODEL, temperature=0),
    tools=[],
    system_prompt="You are a concise support assistant.",
    checkpointer=InMemorySaver(),
)
config = {"configurable": {"thread_id": "customer-001"}}
````

## 3. Store the first message

````python
first = agent.invoke(
    {"messages": [{"role": "user", "content": "I use the Business plan."}]},
    config=config,
)
print(first["messages"][-1].content)
````

## 4. Ask a follow-up question

````python
second = agent.invoke(
    {"messages": [{"role": "user", "content": "Which plan did I say I use?"}]},
    config=config,
)
print(second["messages"][-1].content)
````

## Example Output

```text
You said you use the Business plan.
```

The exact wording may vary for model-generated responses, while the expected facts and structure should remain consistent.

## Think Summary

### How does thread-level memory work?

- The checkpointer stores messages and state after each invocation.
- The same thread_id reloads the same conversation history.
- A different thread_id creates an independent conversation.
- InMemorySaver is suitable for demos; production needs persistent storage.

### Key takeaway

InMemorySaver is suitable for demos; production needs persistent storage.
