## 1. Import LCEL components

````python
from langchain.chat_models import init_chat_model
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv

load_dotenv()
MODEL = "openai:gpt-4o-mini"
````

## 2. Compose the chain

````python
prompt = ChatPromptTemplate.from_messages([
    ("system", "Summarize support tickets in one sentence."),
    ("human", "{ticket}"),
])
model = init_chat_model(MODEL, temperature=0)
chain = prompt | model | StrOutputParser()
````

## 3. Invoke the chain

````python
result = chain.invoke({
    "ticket": "The customer cannot log in after enabling SSO."
})
print(result)
````

## Example Output

```text
The customer cannot log in after enabling single sign-on.
```

The exact wording may vary for model-generated responses, while the expected facts and structure should remain consistent.

## Think Summary

### Why is this better implemented as a chain rather than an agent?

- The workflow has a known, fixed sequence.
- No dynamic tool selection or iterative reasoning is required.
- A chain is simpler, faster, cheaper, and easier to test for deterministic transformations.

### Key takeaway

A chain is simpler, faster, cheaper, and easier to test for deterministic transformations.
