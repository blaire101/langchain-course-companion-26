## 1. Initialize the model

````python
from typing import Literal
from pydantic import BaseModel, Field
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv

load_dotenv()
MODEL = "openai:gpt-4o-mini"
````

## 2. Define the output schema

````python
class SupportTicket(BaseModel):
    category: Literal["billing", "technical", "product", "other"]
    priority: Literal["low", "medium", "high"]
    summary: str = Field(description="One-sentence summary")
````

## 3. Bind the schema

````python
model = init_chat_model(MODEL, temperature=0)
structured_model = model.with_structured_output(SupportTicket)
````

## 4. Invoke the structured model

````python
ticket = "Our production API is down and all users are affected."
result = structured_model.invoke(
    f"Classify this support ticket: {ticket}"
)
result
````

## Example Output

```text
SupportTicket(category='technical', priority='high', summary='Production API outage affecting all users.')
```

The exact wording may vary for model-generated responses, while the expected facts and structure should remain consistent.

## Think Summary

### Why is a validated Pydantic object safer than plain-text JSON?

- Pydantic validates required fields and allowed values.
- The application receives a typed object instead of parsing unpredictable text.
- Validation failures are visible and can be handled explicitly.

### Key takeaway
