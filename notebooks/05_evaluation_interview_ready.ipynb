## 1. Define evaluation cases

````python
tests = [
    {"question": "Which plan has audit logs?", "expected": "Business"},
    {"question": "What is the new-customer refund window?", "expected": "fourteen"},
    {"question": "What is the Priority 1 response time?", "expected": "thirty minutes"},
]
````

## 2. Create a baseline scorer

````python
def keyword_score(answer: str, expected: str) -> int:
    """Return 1 when the expected keyword is present, otherwise 0."""
    return int(expected.lower() in answer.lower())
````

## 3. Score sample answers

````python
sample_answers = [
    "The Business plan includes audit logs.",
    "The refund window is fourteen calendar days.",
    "Priority 1 receives a response within thirty minutes.",
]

scores = [
    keyword_score(answer, test["expected"])
    for answer, test in zip(sample_answers, tests)
]

average_score = sum(scores) / len(scores)
print(average_score)
````

## Think Summary

### Why is keyword scoring only a baseline?

- A keyword may appear in an incorrect or unsupported answer.
- A correct paraphrase may not contain the exact expected word.
- Production evaluation should measure retrieval, groundedness, correctness, relevance, and format separately.
