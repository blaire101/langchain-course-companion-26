# 2. Memory
Use a LangGraph checkpointer to preserve messages by thread.

```python
from langchain.agents import create_agent
from langchain.chat_models import init_chat_model
from langgraph.checkpoint.memory import InMemorySaver
from dotenv import load_dotenv
load_dotenv()
```

```python
agent = create_agent(
    model=init_chat_model('openai:gpt-4.1-mini'),
    tools=[],
    system_prompt='You are a concise support assistant.',
    checkpointer=InMemorySaver(),
)
config = {'configurable': {'thread_id': 'customer-001'}}
```

```python
agent.invoke({'messages':[{'role':'user','content':'I use the Business plan.'}]}, config=config)
```

```python
reply = agent.invoke({'messages':[{'role':'user','content':'Which plan did I say I use?'}]}, config=config)
reply['messages'][-1].content
```

## Production note
In-memory state is suitable only for a demo. Use a persistent checkpointer in production.
